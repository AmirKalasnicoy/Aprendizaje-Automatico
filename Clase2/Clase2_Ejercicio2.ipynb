```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
autos = pd.read_csv("Automobile.csv")
autos.head()
```

```python
plt.scatter(autos["mpg"], autos["weight"])
plt.xlabel("MPG")
plt.ylabel("Weight")
plt.title("Relación entre MPG y Weight")
plt.show()
```

```python
plt.hist(autos["mpg"])
plt.xlabel("MPG")
plt.ylabel("Frecuencia")
plt.title("Distribución del consumo de combustible")
plt.show()
```

```python
autos.boxplot(column="cylinders", by="origin")
plt.xlabel("Origen")
plt.ylabel("Cilindros")
plt.title("Distribución de cilindros según origen")
plt.suptitle("")
plt.show()
```

```python
tabla_barras = autos.groupby(["model_year", "origin"]).size().unstack()

tabla_barras.plot(kind="bar")

plt.xlabel("Año del modelo")
plt.ylabel("Cantidad de autos")
plt.title("Distribución de autos por año y origen")
plt.show()
```
